```python
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)
snapshot_date
```

```python
last_purchase = df.groupby('CustomerID')['InvoiceDate'].max()
recency = (snapshot_date - last_purchase).dt.days
recency.head()
```

```python
frequency = df.groupby('CustomerID')['InvoiceNo'].nunique()
frequency.head()
```

```python
monetary = df.groupby('CustomerID')['TotalPrice'].sum()
monetary.head()
```

```python
rfm =  pd.DataFrame({'Recency':recency,
                     'Frequency':frequency,
                     'Monetary':monetary}).reset_index()
print(rfm.shape)
rfm.head()
```

```python
rfm.describe()
```

```python
rfm.to_csv('/content/drive/MyDrive/retail-rfm-project/rfm_table.csv', index=False)
```

```python
rfm['R_score'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1]).astype(int)
```

```python
rfm['F_score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)
```

```python
rfm['M_score'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5]).astype(int)
```

```python
rfm.head()
```

```python
rfm['RFM_score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)
rfm.head()
```

```python
conditions = [
    (rfm['R_score'] >= 4) & (rfm['F_score'] >= 4),   # Champions
    (rfm['R_score'] <= 2) & (rfm['F_score'] >= 3),   # At Risk
    (rfm['R_score'] <= 2) & (rfm['F_score'] <= 2)    # Lost
]
names = ['Champions', 'At Risk', 'Lost']

rfm['Segment'] = np.select(conditions, names, default='Loyal')
```

```python
rfm['Segment'].value_counts()
```

```python
segment_summary = rfm.groupby('Segment').agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean'),
    Total_Revenue=('Monetary', 'sum')
).round(1)

segment_summary['Revenue_%'] = (segment_summary['Total_Revenue'] / segment_summary['Total_Revenue'].sum() * 100).round(1)
segment_summary['Customers_%'] = (segment_summary['Customers'] / segment_summary['Customers'].sum() * 100).round(1)

segment_summary
```

```python
rfm.to_csv('/content/drive/MyDrive/retail-rfm-project/rfm_segments.csv', index=False)
segment_summary.to_csv('/content/drive/MyDrive/retail-rfm-project/segment_summary.csv')
```

```python
roi = segment_summary[['Customers', 'Avg_Frequency', 'Avg_Monetary']].copy()
roi
```

```python
roi['Avg_Order_Value'] = roi['Avg_Monetary'] / roi['Avg_Frequency']
roi['Avg_Order_Value']
```

```python
roi['Response_Rate'] = [0.12 , 0.05, 0.02, 0.10]
roi['Cost_per_Customer'] = [6, 3, 6, 3]
roi
```

```python
roi['Campaign_cost'] = roi['Customers'] * roi['Cost_per_Customer']
roi['Expected_Revenue'] = roi['Customers'] * roi['Response_Rate'] * roi['Avg_Order_Value']
roi[['Campaign_cost', 'Expected_Revenue']].round(0)
```

```python
gross_margin = 0.30
roi['Expected_Profit'] = roi['Expected_Revenue'] * gross_margin
roi['Net_Gain'] = roi['Expected_Profit'] - roi['Campaign_cost']
roi['ROI'] = roi['Net_Gain'] / roi['Campaign_cost']
roi[['Expected_Profit', 'Net_Gain', 'ROI']].round(2)
```

```python
total_budget = 10000
roi['Gain_For_Budget'] = roi['Net_Gain'].clip(lower=0)
roi['Budget_Share_%'] = roi['Gain_For_Budget'] / roi['Gain_For_Budget'].sum() * 100
roi['Budget_GBP'] = roi['Budget_Share_%'] / 100 * total_budget
roi[['Budget_Share_%', 'Budget_GBP']].round(1)

```
